# National Assembly of Pakistan Dataset
**This Dataset contains the information about the members of National Assembly of Pakistan.**

Today we will get to know more about the politicans from data.
**Let's begin!**

## 1- Importing Libraries 

```python

import pandas as pd

import numpy as np

import matplotlib.pyplot as plt

import seaborn as sns
#To give the graph some dark color.
sns.set_style('darkgrid')
```

### 2- Loading Data

```python
# Use the below line if you run into an error
#pip install openpyxl
```

```python
df = pd.read_excel('/kaggle/input/national-assembly-of-pakistan/NA_list.xlsx')
df.head()
```

## 3-Data Exploration Begins!!!

Let's check the shape of the data.

```python
df.shape
```

**So, we our dataset contains data about 340 Politicans.**

```python
df.info()
```

**Lets check for null values**

```python
df.isnull().sum()
```

**There are some null values in the contact column.Let's just drop the whole column, since its not useful for us.**

```python
df=df.drop('Contact',axis=1)
```

```python
sns.histplot(df['Party'])
plt.ylabel('Number of candidates');
```

**PTI got the majority in the National Assembly.**

```python
candidates_per_party=df.groupby(['Party'])['Party'].count().sort_values(ascending=False)
candidates_per_party
```

**Top 3 Political Parties of Pakistan are occupying 87% seats in the National Assembly.**

### Plotting the above results

```python
plt.figure(figsize=(15, 9))
sns.barplot(y=candidates_per_party.index,x=candidates_per_party)
plt.xlabel('Candidates in the National Assembly of Pak')
;
```

### Let's move on to their education.

```python
education = df.groupby('Profession/Education')['Party'].count().sort_values(ascending=False)
education
```

```python
plt.figure(figsize=(15, 9))
sns.barplot(y=education.index[:10],x=education[:10])
plt.xlabel('Candidates in the National Assembly of Pak')
;
```

## Education per party

```python
Education_party=df.groupby(['Party','Profession/Education'])['Party'].count().sort_values(ascending=False).unstack(level=1).fillna(0)
Education_party
```

# Conclusion


* PTI have the majority in the National Assembly with 155 seats.
* Top 3 Political Parties of Pakistan(PTI,PMLN and PPP) are occupying 87% seats in the National Assembly.
* Most of the People in National Assembly have completed their graduation.
